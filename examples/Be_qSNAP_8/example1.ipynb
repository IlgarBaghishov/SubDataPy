```python
import subdatapy.data as sd
import subdatapy.subsampler.random as sr
import subdatapy.subsampler.leverage as sl
import subdatapy.subsampler.cooks as sc
import numpy as np

X = np.load("X.npy")
y = np.load("y.npy")
w = np.load("w.npy")
config_idxs = np.load("config_idxs.npy")
# enrow_mask = np.load("enrow_mask.npy")
```

```python
bd = sd.BaseData(X,y,w,config_idxs)
bd.train_test_split(test_fraction=0.5, seed=42)
bd.train()
bd.compute_errors()
```

```python
rs = sr.RandomSubSampler(X,y,w,test_fraction=0.5,seed=42,config_idxs=config_idxs)
rs.create_subsample(subsample_fraction=0.2)
rs.train_subsample()
rs.compute_subsample_errors()
```

```python
ls = sl.LeverageSubSampler(X,y,w,test_fraction=0.5,seed=42,config_idxs=config_idxs)
ls.create_subsample(subsample_fraction=0.2)
ls.train_subsample()
ls.compute_subsample_errors()
```

```python
cs = sc.CookSubSampler(X, y, w, test_fraction=0.5, seed=42, config_idxs=config_idxs)
cs.create_subsample(subsample_fraction=0.2)
cs.train_subsample()
cs.compute_subsample_errors()
```

```python
rs = sr.RandomSubSampler(X,y,w,test_fraction=0.5,seed=42,config_idxs=config_idxs)
errors_df = rs.create_subsample_errors_dataframe(subsample_fractions_list=[0.2, 0.4, 0.6, 0.8], repeat_count_list=[20, 15, 10, 5])  # use an int instead of a list for repeat_count_list to repeat the same number of times for all subsample_fractions
errors_df.describe()
```
